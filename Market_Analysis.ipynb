```python
# =========================================================
# COMPLETE DATA ANALYSIS PROJECT
# Dataset : market_3.csv
# =========================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# -----------------------------
# 2. LOAD DATASET
# -----------------------------
df = pd.read_csv("C:/Users/arfat/OneDrive/ドキュメント/Data_Science/Data_sets/market_3.csv")
```

```python
# -----------------------------
# 3. DISPLAY FIRST & LAST ROWS
# -----------------------------
print("\n================ FIRST 5 ROWS ================\n")
print(df.head())

print("\n================ LAST 5 ROWS ================\n")
print(df.tail())
```

```python
# -----------------------------
# 4. DATASET SHAPE
# -----------------------------
print("\n================ DATASET SHAPE ================\n")
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])
```

```python
# -----------------------------
# 5. COLUMN NAMES
# -----------------------------
print("\n================ COLUMN NAMES ================\n")
print(df.columns)
```

```python
# -----------------------------
# 6. DATA TYPES
# -----------------------------
print("\n================ DATA TYPES ================\n")
print(df.dtypes)
```

```python
# -----------------------------
# 7. DATASET INFO
# -----------------------------
print("\n================ DATASET INFO ================\n")
print(df.info())
```

```python
# -----------------------------
# 8. STATISTICAL SUMMARY
# -----------------------------
print("\n================ STATISTICAL SUMMARY ================\n")
print(df.describe())
```

```python
# -----------------------------
# 9. MISSING VALUES
# -----------------------------
print("\n================ MISSING VALUES ================\n")
print(df.isnull().sum())

print("\nTotal Missing Values :",
      df.isnull().sum().sum())
```

```python
# -----------------------------
# 10. DUPLICATE VALUES
# -----------------------------
print("\n================ DUPLICATE VALUES ================\n")
print("Duplicate Rows :", df.duplicated().sum())
```

```python
# -----------------------------
# 11. IDENTIFY NUMERIC & CATEGORICAL COLUMNS
# -----------------------------
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(exclude=np.number).columns

print("\nNumerical Columns :\n", num_cols)
print("\nCategorical Columns :\n", cat_cols)
```

```python
# =========================================================
# VISUALIZATION SECTION
# =========================================================

# -----------------------------
# 12. HISTOGRAMS
# -----------------------------
df[num_cols].hist(figsize=(14,10))

plt.suptitle("Histogram of Numerical Features")

plt.tight_layout()

plt.show()
```

```python
# -----------------------------
# 13. BOXPLOTS
# -----------------------------
for col in num_cols:

    plt.figure(figsize=(6,4))

    plt.boxplot(df[col].dropna())

    plt.title(f"Boxplot of {col}")

    plt.ylabel(col)

    plt.show()
```

```python
# -----------------------------
# 14. LINE PLOTS
# -----------------------------
for col in num_cols:

    plt.figure(figsize=(8,4))

    plt.plot(df[col])

    plt.title(f"Line Plot of {col}")

    plt.xlabel("Index")

    plt.ylabel(col)

    plt.show()
```

```python
# -----------------------------
# 15. BAR CHARTS
# -----------------------------
for col in cat_cols:

    plt.figure(figsize=(8,5))

    df[col].value_counts().plot(kind='bar')

    plt.title(f"Bar Chart of {col}")

    plt.xlabel(col)

    plt.ylabel("Count")

    plt.show()
```

```python
# -----------------------------
# 16. PIE CHARTS
# -----------------------------
for col in cat_cols:

    plt.figure(figsize=(6,6))

    df[col].value_counts().plot(
        kind='pie',
        autopct='%1.1f%%'
    )

    plt.ylabel("")

    plt.title(f"Pie Chart of {col}")

    plt.show()
```

```python
# -----------------------------
# 17. SCATTER PLOTS
# -----------------------------
if len(num_cols) >= 2:

    for i in range(len(num_cols)-1):

        plt.figure(figsize=(7,5))

        plt.scatter(
            df[num_cols[i]],
            df[num_cols[i+1]]
        )

        plt.xlabel(num_cols[i])

        plt.ylabel(num_cols[i+1])

        plt.title(
            f"{num_cols[i]} vs {num_cols[i+1]}"
        )

        plt.show()
```

```python
# -----------------------------
# 18. CORRELATION MATRIX
# -----------------------------
corr = df[num_cols].corr()

plt.figure(figsize=(10,8))

plt.imshow(corr)

plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("Correlation Matrix")

plt.show()

print("\n================ CORRELATION MATRIX ================\n")
print(corr)
```

```python
# -----------------------------
# 19. OUTLIER DETECTION
# -----------------------------
print("\n================ OUTLIER DETECTION ================\n")

for col in num_cols:

    Q1 = df[col].quantile(0.25)

    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)

    upper = Q3 + (1.5 * IQR)

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(f"{col} : {len(outliers)} outliers")
```

```python
# -----------------------------
# 20. MEAN, MEDIAN, MODE
# -----------------------------
print("\n================ CENTRAL TENDENCY ================\n")

for col in num_cols:

    print(f"\nColumn : {col}")

    print("Mean   :", df[col].mean())

    print("Median :", df[col].median())

    print("Mode   :", df[col].mode()[0])
```

```python
# -----------------------------
# 21. STANDARD DEVIATION & VARIANCE
# -----------------------------
print("\n================ DISPERSION ================\n")

for col in num_cols:

    print(f"\nColumn : {col}")

    print("Standard Deviation :", df[col].std())

    print("Variance           :", df[col].var())
```

```python
# -----------------------------
# 22. SKEWNESS & KURTOSIS
# -----------------------------
print("\n================ DISTRIBUTION ANALYSIS ================\n")

for col in num_cols:

    print(f"\nColumn : {col}")

    print("Skewness :", df[col].skew())

    print("Kurtosis :", df[col].kurt())
```

```python
# -----------------------------
# 23. UNIQUE VALUES
# -----------------------------
print("\n================ UNIQUE VALUES ================\n")

for col in cat_cols:

    print(f"\n{col}")

    print(df[col].unique())
```

```python
# -----------------------------
# 24. VALUE COUNTS
# -----------------------------
print("\n================ VALUE COUNTS ================\n")

for col in cat_cols:

    print(f"\n{col}")

    print(df[col].value_counts())
```

```python
# -----------------------------
# 25. GROUP ANALYSIS
# -----------------------------
if len(cat_cols) > 0 and len(num_cols) > 0:

    for cat in cat_cols:

        print(f"\n======= GROUP ANALYSIS : {cat} =======\n")

        print(df.groupby(cat)[num_cols[0]].mean())
```

```python
# -----------------------------
# 26. SAVE CLEANED DATASET
# -----------------------------
df.to_csv(
    r"C:\Users\arfat\Downloads\Github-Code\Cleaned_market_data.csv",
    index=False
)

print("\nCleaned Dataset Saved Successfully!")

# =========================================================
# END OF PROJECT
# =========================================================
```
